### Environment Setup

Create the `dev` catalog and `dev.stripex` schema, the `landing` volume, and its 7 subfolders.

```python
%sql
-- Create the dev catalog
CREATE CATALOG IF NOT EXISTS dev;
```

```python
%sql
-- Create dev,stripex schema 
CREATE SCHEMA IF NOT EXISTS dev.stripex;
```

```python
%sql
-- Create landing zone volume in dev.stripex volume
CREATE VOLUME IF NOT EXISTS dev.stripex.landing;
```

### Create the 7 landing subfolders

Three CDC subfolders (orders, order_items, customers) and four file-based subfolders (products, categories, clickstream, inventory), 1 folder per source, inside `landing` volume.

```python
landing_root = "/Volumes/dev/stripex/landing"

subfolders = [
    "orders_cdc",
    "order_items_cdc",
    "customers_cdc",
    "products",
    "categories",
    "clickstream",
    "inventory",
]

for folder in subfolders:
    dbutils.fs.mkdirs(f"{landing_root}/{folder}")
    print(f"Created: {landing_root}/{folder}")
```
