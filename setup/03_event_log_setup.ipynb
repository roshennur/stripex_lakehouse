### DQ Monitoring - Event Log View Setup

Creates the `event_log_raw` view for each pipeline. \
**Constraint:** the event_log table-valued function can not be used to access the event logs of multiple pipelines.

Two pipelines, two seperate views: `ingestion_event_log_raw` and `transformation_event_log_raw`

Only the pipeline owner can create this view, and it cant't be shared with other users.

```python
dbutils.widgets.text("catalog", "dev", "Target catalog")
dbutils.widgets.text("ingestion_pipeline_id", "8284e955-8eee-4ee0-97d9-3a78b1541d40", "stripex-ingestion-pipeline ID")
dbutils.widgets.text("transformation_pipeline_id", "521e251c-453a-4b50-b1e5-398e75fdc7a3", "stripex-transformation-pipeline ID")

catalog = dbutils.widgets.get("catalog")
ingestion_pipeline_id = dbutils.widgets.get("ingestion_pipeline_id")
transformation_pipeline_id = dbutils.widgets.get("transformation_pipeline_id")

if not ingestion_pipeline_id or not transformation_pipeline_id:
    raise ValueError(
        "Both pipeline ID's are required."    
    )
```

```python
spark.sql(f"""
CREATE OR REPLACE VIEW {catalog}.stripex.ingestion_event_log_raw AS
SELECT * FROM event_log("{ingestion_pipeline_id}")
""")

spark.sql(f"""
CREATE OR REPLACE VIEW {catalog}.stripex.transformation_event_log_raw AS
SELECT * FROM event_log("{transformation_pipeline_id}")
""")

print(f"Created {catalog}.stripex.ingestion_event_log_raw")
print(f"Created {catalog}.stripex.transformation_event_log_raw")
```

```python
%sql
SELECT * FROM dev.stripex.ingestion_event_log_raw
LIMIT 10;
```
